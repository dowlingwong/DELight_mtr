```python
from tcxformer_v2 import *
if __name__ == "__main__":
    torch.manual_seed(0)
    cfg = ModelConfig(
        n_photon=19,
        n_phonon=37,
        stride_photon=128,
        stride_phonon=256,
        d_model=256,
        d_ff=1024,
        n_heads=4,
        n_time_layers=3,
        n_chan_layers=1,
        patch_embed=64,
        dropout=0.1,
        rope_base=10000.0,
        n_branch_to_task=2,
    )
    model = ModalityTCXFormer(cfg)
    model.eval()

    B, C, L = 2, 56, 65536
    x = torch.randn(B, C, L)
    mask = torch.ones(B, C)
    with torch.no_grad():
        pos, energy, cls_logit, aux = model(x, channel_mask=mask)
    print("pos:", pos.shape, "energy:", energy.shape, "cls:", cls_logit.shape)
    print({k: v.shape for k, v in aux.items()})
```
