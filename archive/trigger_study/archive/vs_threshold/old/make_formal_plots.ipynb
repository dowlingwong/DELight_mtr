```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
import h5py

# --- Settings ---
base_root = Path("/ceph/dwong/work/threshold/pink_run1")
noise_powers = [1.75, 6.73, 16.60, 69.13]

A_SIGMA = 0.162217  # fallback: sigma(noise_power) ≈ A * sqrt(noise_power)

# Sliding-fit sigmas from determine_power.ipynb (pink noise only)
PINK_SIGMA_QP = {
    1.75: 0.581095,
    6.73: 1.19541,
    16.60: 1.81300,
    69.13: 3.70415,
}
PINK_SIGMA_UV = {
    1.75: 0.469716,
    6.73: 0.941077,
    16.60: 1.45092,
    69.13: 2.96408,
}
SIGMA_TEMPLATE = "uv"  # "uv" or "qp"

# We always want 5σ, 6σ, 7σ
sigma_multipliers = [5, 6, 7]
mult_labels = {5: "5σ", 6: "6σ", 7: "7σ"}
energy_re = re.compile(r"amplitudes_energy_(?P<E>[-+]?\d+(?:\.\d+)?)\.h5$")

TARGET_PASS = 0.5  # detection threshold at 50% pass
POLY_DEG = 2       # polynomial degree for smooth fit on threshold vs resolution

PLOT_COLORS = {5: "#1f77b4", 6: "#ff7f0e", 7: "#2ca02c"}
PLOT_MARKERS = {5: "o", 6: "s", 7: "^"}


def _noise_dir(noise_power: float) -> Path:
    return base_root / f"noise_{noise_power:.1f}"


def load_amplitudes(path: Path) -> np.ndarray:
    with h5py.File(path, "r") as f:
        if "amplitudes" not in f:
            raise KeyError(f"Missing 'amplitudes' dataset in {path}")
        data = np.asarray(f["amplitudes"][:])
        n_repeats = f.attrs.get("n_repeats")
        n_channels = f.attrs.get("n_channels")
    if data.ndim == 1:
        if n_channels == data.shape[0]:
            data = data[None, :]
        elif n_repeats == data.shape[0]:
            data = data[:, None]
        else:
            data = data[None, :]
    if data.ndim != 2:
        raise ValueError(
            f"Unexpected shape in {path}: {data.shape} (need (repeats, n_channels))"
        )
    return data


def sigma_from_sliding_fit(noise_power: float) -> tuple[float, str]:
    sigma_map = PINK_SIGMA_UV if SIGMA_TEMPLATE == "uv" else PINK_SIGMA_QP
    key = round(float(noise_power), 2)
    if key in sigma_map:
        return float(sigma_map[key]), f"sliding fit ({SIGMA_TEMPLATE})"
    return float(A_SIGMA * np.sqrt(noise_power)), "formula"


def _interp_threshold_energy(energies, rates, target):
    energies = np.asarray(energies, dtype=float)
    rates = np.asarray(rates, dtype=float)
    mask = np.isfinite(energies) & np.isfinite(rates)
    energies = energies[mask]
    rates = rates[mask]
    if energies.size < 2:
        return np.nan
    order = np.argsort(energies)
    energies = energies[order]
    rates = rates[order]
    for i, rate in enumerate(rates):
        if rate >= target:
            if i == 0:
                return float(energies[0])
            e0, e1 = energies[i - 1], energies[i]
            r0, r1 = rates[i - 1], rates[i]
            if r1 == r0:
                return float(e1)
            return float(e0 + (target - r0) * (e1 - e0) / (r1 - r0))
    return np.nan


# Precompute sigmas from sliding-fit results
noise_sigmas = {p: sigma_from_sliding_fit(p) for p in noise_powers}

threshold_energies = {m: [] for m in sigma_multipliers}
resolution_sigmas = []

for noise_power in noise_powers:
    base_dir = _noise_dir(noise_power) / "fit_with_shift_amp"
    if not base_dir.exists():
        print(f"[warn] Missing noise directory: {base_dir}")
        continue

    files = []
    for f in base_dir.glob("amplitudes_energy_*.h5"):
        m = energy_re.match(f.name)
        if m:
            files.append((float(m.group("E")), f))
    files.sort(key=lambda t: t[0])

    if not files:
        print(f"[warn] No amplitude files found in {base_dir}")
        continue

    sigma, sigma_src = noise_sigmas[noise_power]
    thresholds = {m: m * sigma for m in sigma_multipliers}

    print()
    print(f"=== Noise power {noise_power} ===")
    print(f"  σ ≈ {sigma:.4f} ({sigma_src})")
    for m in sigma_multipliers:
        print(f"  {m}σ threshold ≈ {thresholds[m]:.4f}")

    pass_rates = {m: [] for m in sigma_multipliers}
    available_E = []
    n_channels = None

    for E, path in files:
        try:
            data = load_amplitudes(path)
        except Exception as exc:
            print(f"[warn] Failed to load {path}: {exc}")
            continue

        if n_channels is None:
            n_channels = data.shape[1]
        elif data.shape[1] != n_channels:
            print(f"[warn] Channel count changed {n_channels} -> {data.shape[1]} for {path}")
            n_channels = data.shape[1]

        available_E.append(E)

        for m in sigma_multipliers:
            thr = thresholds[m]
            passes = (data >= thr).any(axis=1)  # per-trace pass across channels
            pass_rates[m].append(passes.mean())

    if not available_E:
        print(f"[warn] No energies found for noise={noise_power}, skipping plot.")
        continue

    plt.figure(figsize=(6.5, 4.5))
    for m in sigma_multipliers:
        plt.plot(available_E, pass_rates[m], label=mult_labels[m])

    plt.xlabel("Energy (eV)")
    plt.ylabel("Pass rate")
    plt.title(f"Pass rate vs Energy (noise = {noise_power})")
    plt.ylim(0.0, 1.05)
    plt.grid(True, which="both", linestyle="--", alpha=0.4)
    plt.legend(title="Thresholds")
    plt.tight_layout()
    plt.show()

    print()
    print("Pass rates:")
    header = "Energy".ljust(8) + " " + "  ".join([f"{mult_labels[m]:>10}" for m in sigma_multipliers])
    print(header)
    for i, E in enumerate(available_E):
        row = [f"{E:<8}"]
        for m in sigma_multipliers:
            row.append(f"{pass_rates[m][i]:>10.4f}")
        print(" ".join(row))
    if n_channels is not None:
        print(f"Channels per trace: {n_channels}")

    resolution_sigmas.append(sigma)
    for m in sigma_multipliers:
        thr_e = _interp_threshold_energy(available_E, pass_rates[m], TARGET_PASS)
        threshold_energies[m].append(thr_e)

if resolution_sigmas:
    plt.figure(figsize=(7.0, 4.8))
    for m in sigma_multipliers:
        color = PLOT_COLORS[m]
        marker = PLOT_MARKERS[m]
        x_all = np.asarray(resolution_sigmas, dtype=float)
        y_all = np.asarray(threshold_energies[m], dtype=float)
        mask = np.isfinite(x_all) & np.isfinite(y_all)
        x = x_all[mask]
        y = y_all[mask]

        plt.scatter(
            x, y,
            color=color,
            marker=marker,
            s=55,
            edgecolors="white",
            linewidths=0.7,
            label=f"{mult_labels[m]} points",
            zorder=3,
        )

        if x.size >= 2:
            deg = min(POLY_DEG, x.size - 1)
            coeffs = np.polyfit(x, y, deg)
            x_fit = np.linspace(x.min(), x.max(), 200)
            y_fit = np.polyval(coeffs, x_fit)
            plt.plot(
                x_fit, y_fit,
                color=color,
                linewidth=2.0,
                label=f"{mult_labels[m]} fit (deg {deg})",
                zorder=2,
            )

    plt.xlabel("Energy resolution σ (eV)")
    plt.ylabel(f"Threshold energy at {TARGET_PASS:.0%} pass (eV)")
    plt.title("Threshold vs Energy Resolution")
    plt.grid(True, which="both", linestyle="--", alpha=0.35)
    plt.legend(ncol=2, frameon=False)
    plt.tight_layout()
    plt.show()
```

```python
# 4-sigma threshold pass-rate curves for all resolutions (pink)
threshold_sigma = 6.0

colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]
markers = ["o", "s", "^", "D"]

legend_labels = [
    r"$\Delta E_{\mathrm{FWHM}} = 1.5\,\mathrm{eV}$",
    r"$\Delta E_{\mathrm{FWHM}} = 2.5\,\mathrm{eV}$",
    r"$\Delta E_{\mathrm{FWHM}} = 4.3\,\mathrm{eV}$",
    r"$\Delta E_{\mathrm{FWHM}} = 8.3\,\mathrm{eV}$",
]

smooth_points = 300
smooth_window = 17  # odd length for Gaussian kernel
smooth_sigma = 6.0  # kernel width in samples


def smooth_curve(x, y, n_points=300, window=11, sigma=2.0):
    order = np.argsort(x)
    x_sorted = x[order]
    y_sorted = y[order]

    x_dense = np.linspace(x_sorted.min(), x_sorted.max(), n_points)
    y_dense = np.interp(x_dense, x_sorted, y_sorted)

    if window % 2 == 0:
        window += 1
    window = max(window, 3)

    t = np.arange(window) - (window - 1) / 2
    kernel = np.exp(-0.5 * (t / sigma) ** 2)
    kernel /= kernel.sum()

    pad = window // 2
    y_pad = np.pad(y_dense, pad_width=pad, mode="edge")
    y_smooth = np.convolve(y_pad, kernel, mode="valid")
    y_smooth = np.clip(y_smooth, 0.0, 1.0)
    return x_dense, y_smooth


plt.figure(figsize=(7.2, 4.8))

for idx, noise_power in enumerate(noise_powers):
    base_dir = _noise_dir(noise_power) / "fit_with_shift_amp"
    if not base_dir.exists():
        print(f"[warn] Missing noise directory: {base_dir}")
        continue

    files = []
    for f in base_dir.glob("amplitudes_energy_*.h5"):
        m = energy_re.match(f.name)
        if m:
            files.append((float(m.group("E")), f))
    files.sort(key=lambda t: t[0])

    if not files:
        print(f"[warn] No amplitude files found in {base_dir}")
        continue

    sigma, sigma_src = sigma_from_sliding_fit(noise_power)
    thr = threshold_sigma * sigma

    energies = []
    pass_rates = []

    for E, path in files:
        try:
            data = load_amplitudes(path)
        except Exception as exc:
            print(f"[warn] Failed to load {path}: {exc}")
            continue

        energies.append(E)
        passes = (data >= thr).any(axis=1)  # per-trace pass across channels
        pass_rates.append(passes.mean())

    if not energies:
        continue

    x = np.asarray(energies, dtype=float)
    y = np.asarray(pass_rates, dtype=float)

    color = colors[idx % len(colors)]
    marker = markers[idx % len(markers)]
    if idx < len(legend_labels):
        label = legend_labels[idx]
    else:
        label = f"noise={noise_power:.2f}"

    plt.scatter(
        x, y,
        color=color,
        marker=marker,
        s=28,
        edgecolors="white",
        linewidths=0.6,
        zorder=3,
    )

    x_smooth, y_smooth = smooth_curve(
        x,
        y,
        n_points=smooth_points,
        window=smooth_window,
        sigma=smooth_sigma,
    )
    plt.plot(
        x_smooth, y_smooth,
        color=color,
        linewidth=1.8,
        label=label,
        zorder=2,
    )

plt.xlabel("Energy of NR envent (eV)")
plt.ylabel("Pass rate(>1 LAMCAL triggered)")
plt.title(f"Pass rate vs Energy (threshold = {threshold_sigma}σ)")
plt.ylim(0.0, 1.05)
plt.grid(True, which="both", linestyle="--", alpha=0.35)
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import h5py

# Gaussian sigma from positive-only noise amplitudes (energy=0)
base_root = Path("/ceph/dwong/work/threshold/pink_run1")
noise_powers = [1.75, 6.73, 16.60, 69.13]

for noise_power in noise_powers:
    path = base_root / f"noise_{noise_power:.1f}" / "fit_with_shift_amp" / "amplitudes_energy_0.h5"
    if not path.exists():
        print(f"[warn] Missing noise= {noise_power} file: {path}")
        continue

    with h5py.File(path, "r") as f:
        if "amplitudes" not in f:
            print(f"[warn] No 'amplitudes' dataset in {path}")
            continue
        amps = np.asarray(f["amplitudes"][:])

    flat = amps.ravel().astype(float)
    pos = flat[flat > 0]
    if pos.size == 0:
        print(f"[warn] No positive amplitudes for noise={noise_power}")
        continue

    mu = float(np.mean(pos))
    sigma = float(np.std(pos, ddof=0))  # MLE sigma on positive side

    print(f"noise={noise_power}: mu≈{mu:.4f}, sigma≈{sigma:.4f}, samples={pos.size}")

    plt.figure(figsize=(6, 4))
    n, bins, _ = plt.hist(pos, bins=50, density=True, alpha=0.6, color="steelblue", label="data (+)")

    xs = np.linspace(bins[0], bins[-1], 200)
    pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((xs - mu) / sigma) ** 2)
    plt.plot(xs, pdf, "r-", label=f"Gaussian fit, σ≈{sigma:.4f}")

    plt.title(f"Positive noise-only amplitudes, noise={noise_power}")
    plt.xlabel("Amplitude")
    plt.ylabel("Density")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
