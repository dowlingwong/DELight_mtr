```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re

# --- Settings ---
base_root = Path("/ceph/dwong/work/threshold")

# All noise powers (the "noise_*" parts of your directory names)
noise_powers = [15.4, 33.2, 42.8, 70, 100.0, 126.7, 200.0, 472]

A_SIGMA = 0.162217  # sigma(noise_power) ≈ A * sqrt(noise_power)

# Energies and file template (unchanged)
energies = list(np.arange(0, 61, 2, dtype=int))
file_tmpl = "max_amplitudes_energy_{E}.npy"

# We always want 3σ, 4σ, 5σ
sigma_multipliers = [3, 4, 5]
mult_labels = {3: "3σ", 4: "4σ", 5: "5σ"}

for noise_power in noise_powers:
    # Build directory for this noise level
    base_dir = base_root / f"noise_{noise_power}" / "indiv_max_amp"

    # Compute sigma and thresholds for this noise level
    sigma = A_SIGMA * np.sqrt(noise_power)
    thresholds = {m: m * sigma for m in sigma_multipliers}

    print(f"\n=== Noise power {noise_power} ===")
    print(f"  σ ≈ {sigma:.4f}")
    for m in sigma_multipliers:
        print(f"  {m}σ threshold ≈ {thresholds[m]:.4f}")

    # --- Compute pass rates for this noise level ---
    pass_rates = {m: [] for m in sigma_multipliers}
    available_E = []

    for E in energies:
        path = base_dir / file_tmpl.format(E=E)
        if not path.exists():
            print(f"[warn] Missing file, skipping: {path}")
            continue

        data = np.load(path)  # expected shape (repeats=100, n_traces=56)
        if data.ndim != 2 or data.shape[1] != 56:
            raise ValueError(f"Unexpected shape in {path}: {data.shape} (expected (R, 56))")

        available_E.append(E)

        # For each threshold (3σ, 4σ, 5σ), event passes if any channel in that repeat >= threshold
        for m in sigma_multipliers:
            thr = thresholds[m]
            passes = (data >= thr).any(axis=1)  # boolean length R
            pass_rate = passes.mean()           # fraction in [0, 1]
            pass_rates[m].append(pass_rate)

    if not available_E:
        print(f"[warn] No energies found for noise={noise_power}, skipping plot.")
        continue

    # --- Plot for this noise level ---
    plt.figure(figsize=(6.5, 4.5))
    for m in sigma_multipliers:
        plt.plot(available_E, pass_rates[m], label=mult_labels[m])

    plt.xlabel("Energy (eV)")
    plt.ylabel("Pass rate")
    plt.title(f"Pass rate vs Energy (noise = {noise_power})")
    plt.ylim(0.0, 1.05)
    plt.grid(True, which="both", linestyle="--", alpha=0.4)
    plt.legend(title="Thresholds")
    plt.tight_layout()
    plt.show()

    # --- Print a small table with σ labels for this noise level ---
    print("\nPass rates:")
    header = "Energy".ljust(8) + " " + "  ".join([f"{mult_labels[m]:>10}" for m in sigma_multipliers])
    print(header)
    for i, E in enumerate(available_E):
        row = [f"{E:<8}"]
        for m in sigma_multipliers:
            row.append(f"{pass_rates[m][i]:>10.4f}")
        print(" ".join(row))
```
