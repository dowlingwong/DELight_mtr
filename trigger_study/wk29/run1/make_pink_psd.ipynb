# Pink noise PSD for wk29
Generate a one-sided pink (1/f) PSD for the wk29 OptimumFilter.
Matches template length 32768 and fs=250000 Hz. Normalization follows the NoiseGenerator pink rule with unit noise power (scale can be updated easily).


```python
import numpy as np
from pathlib import Path

# Parameters
fs = 250_000.0
template_len = 32768  # matches wk15 templates
noise_power = 1.0      # normalization target; OF amplitude is scale-invariant
out_path = Path('pink_psd.npy')

# Frequency grid (one-sided)
freqs = np.fft.rfftfreq(template_len, d=1.0/fs)
f_safe = freqs.copy()
if f_safe[0] == 0:
    f_safe[0] = f_safe[1]  # avoid divide by zero

# Pink shape and NoiseGenerator-style normalization
shape = 1.0 / f_safe
normalize = 1.0 / (np.log(freqs.max()) - np.log(freqs[1]))
norm = 0.5 * noise_power * normalize * fs * template_len
psd = norm * shape

np.save(out_path, psd)
print('Saved', out_path.resolve())
print('Length', psd.shape[0])
print('Min/Max', psd.min(), psd.max())
print('First 5 freqs', freqs[:5])
print('First 5 psd', psd[:5])
```
