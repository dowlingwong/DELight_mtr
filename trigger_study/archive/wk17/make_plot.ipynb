```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import RegularPolygon, Patch
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

# --------------------
# Config
# --------------------
sensor_z        = -1692
hex_radius_sel  = 49.5 / 2.0      # base radius for selected (Vacuum)
other_scale     = 1.15            # scale factor to make unselected (Submerged) larger
hex_radius_oth  = hex_radius_sel * other_scale
orientation     = np.pi / 6       # match 2D
circle_r        = 126.635         # mm
zoom_pad        = 60.0            # mm

# Colors
color_sel = "orange"              # Vacuum channels (selected)
color_oth = "royalblue"           # Submerged channels (unselected)

positions = ts.LAMCAL_position  # shape (N, 3)

# Extract x, y, z
x = positions[:, 0]
y = positions[:, 1]
z = positions[:, 2]

# Groups
mask_sel = (z == sensor_z)
mask_oth = ~mask_sel

selected = positions[mask_sel]
others   = positions[mask_oth]

# --------------------
# helpers
# --------------------
def hexagon_vertices_like_2d(xc, yc, zc, radius):
    """Match 2D RegularPolygon orientation/size, then lift to z=zc."""
    rp = RegularPolygon((xc, yc),
                        numVertices=6,
                        radius=radius,
                        orientation=orientation)
    verts2d = rp.get_verts()  # (6,2)
    return [(vx, vy, zc) for vx, vy in verts2d]

# Circle on the selected plane
theta = np.linspace(0, 2*np.pi, 400)
circle_x = circle_r * np.cos(theta)
circle_y = circle_r * np.sin(theta)
circle_z = np.full_like(circle_x, sensor_z)

# --------------------
# plotting
# --------------------
fig = plt.figure(figsize=(20, 20))
ax = fig.add_subplot(111, projection="3d")

# Selected (Vacuum) in orange
for xi, yi, zi in selected:
    verts3d = hexagon_vertices_like_2d(xi, yi, zi, hex_radius_sel)
    ax.add_collection3d(Poly3DCollection([verts3d],
                                         facecolor=color_sel,
                                         edgecolor="k",
                                         alpha=0.8,
                                         linewidths=0.8))

# Others (Submerged) in blue, bigger
for xi, yi, zi in others:
    verts3d = hexagon_vertices_like_2d(xi, yi, zi, hex_radius_oth)
    ax.add_collection3d(Poly3DCollection([verts3d],
                                         facecolor=color_oth,
                                         edgecolor="k",
                                         alpha=0.6,
                                         linewidths=0.7))

# Circle on selected plane

# Labels / title
ax.set_xlabel("x [mm]", labelpad=10)
ax.set_ylabel("y [mm]", labelpad=10)
ax.set_zlabel("z [mm]", labelpad=15)
ax.set_title("V2 Pancake Geometry")

# Zoom around selected plane footprint
if selected.size > 0:
    sel_xmin, sel_xmax = selected[:,0].min() - zoom_pad, selected[:,0].max() + zoom_pad
    sel_ymin, sel_ymax = selected[:,1].min() - zoom_pad, selected[:,1].max() + zoom_pad
    ax.set_xlim(sel_xmin, sel_xmax)
    ax.set_ylim(sel_ymin, sel_ymax)
else:
    pad = 60.0
    ax.set_xlim(x.min()-pad, x.max()+pad)
    ax.set_ylim(y.min()-pad, y.max()+pad)

# Keep fuller z-range (tweak as needed)
z_pad = 50.0
ax.set_zlim(z.min()-z_pad, z.max()+z_pad)

# Camera angle and aspect
ax.view_init(elev=20, azim=-60)
xr, yr, zr = ax.get_xlim(), ax.get_ylim(), ax.get_zlim()
ax.set_box_aspect((xr[1]-xr[0], yr[1]-yr[0], zr[1]-zr[0]))

# Legend (use 2D proxy patches since 3D collections don't auto-legend)
legend_handles = [
    Patch(facecolor=color_sel, edgecolor='k', label='Vacuum channels'),
    Patch(facecolor=color_oth, edgecolor='k', label='Submerged channels'),
]
ax.legend(handles=legend_handles, loc='upper right', bbox_to_anchor=(1.05, 1.0))

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- mapping: "selected channel label" -> real file index & (x,y) ----------
mask_sel = (ts.LAMCAL_position[:, 2] == sensor_z)
selected_pos = ts.LAMCAL_position[mask_sel]            # shape (Nsel, 3), ordered like your labels
selected_file_idx = np.where(mask_sel)[0]              # real indices into trace array, same order

def ch_to_file_idx(ch_label: int) -> int:
    """Map displayed channel label (19-based) to trace array index."""
    j = ch_label - 19
    if j < 0 or j >= len(selected_file_idx):
        raise IndexError(f"Channel label {ch_label} is out of range for this plane.")
    return int(selected_file_idx[j])

def ch_to_xy(ch_label: int):
    """Get (x,y) for a displayed channel label (19-based)."""
    j = ch_label - 19
    xy = selected_pos[j, :2]
    return float(xy[0]), float(xy[1])

# ---------- groups to plot (center -> neighbors) ----------
groups = {
    37: [43, 36, 30, 31, 38, 44],
    30: [29, 24, 25, 36, 37, 31],
    24: [23, 19, 20, 25, 30, 29],
}

# time axis in ms
times = 0.5 * (ts.t_edges[1:] + ts.t_edges[:-1])
t_ms = times * 1e-6

def plot_center_group(ax, center_label, neighbors):
    # generate a fresh trace with the source at the center channel's (x,y)
    x0, y0 = ch_to_xy(center_label)
    tr = ts.generate(2e2, x0, y0, -1706, type_recoil='NR', phonon_only=True,
                     no_noise=True, quantize=False)

    # helper to plot a single channel (normalize safely)
    def plot_ch(ch_label, *, is_center=False):
        i_real = ch_to_file_idx(ch_label)
        sig = tr[0][i_real]
        peak = np.max(np.abs(sig))
        if peak <= 0:
            return
        if is_center:
            ax.plot(t_ms, sig/peak, lw=2.6, label=f"{ch_label} (center)", color="red")
        else:
            ax.plot(t_ms, sig/peak, lw=1.2, label=str(ch_label))

    # center first (red), then neighbors
    plot_ch(center_label, is_center=True)
    for nb in neighbors:
        plot_ch(nb, is_center=False)

    ax.set_title(f"Centered at ch {center_label}")
    ax.set_xlim(0, np.max(t_ms))
    ax.set_xlabel("Time [ms]")
    ax.set_ylabel("Normalized trace")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper right", fontsize=8, frameon=False)

# ---------- 2×2 subplot assembly ----------
fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True, sharey=True)
axes = axes.ravel()

for ax, (center, nbrs) in zip(axes, groups.items()):
    plot_center_group(ax, center, nbrs)

fig.suptitle("z=-1706", y=0.98)
fig.tight_layout()
plt.show()
```
