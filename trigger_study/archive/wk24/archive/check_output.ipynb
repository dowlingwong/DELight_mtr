```python
import h5py
import pandas as pd
from pathlib import Path

def read_meta_h5(meta_path: Path):
    with h5py.File(meta_path, "r") as f:
        # file-level attributes (handy for shape/dtypes)
        attrs = {k: (v.decode() if isinstance(v, bytes) else v) for k, v in f.attrs.items()}
        data = f["events"][:]   # structured array

    # convert structured array to a tidy DataFrame
    df = pd.DataFrame({
        "x": data["x"],
        "y": data["y"],
        "z": data["z"],
        "energy": data["energy"],
        # 'type_recoil' was stored as HDF5 variable-length UTF-8
        "type_recoil": [s.decode("utf-8") if isinstance(s, (bytes, bytearray)) else str(s) 
                        for s in data["type_recoil"]],
        "no_noise": data["no_noise"],
        "quantize": data["quantize"],
    })
    return attrs, df

# Usage
meta_path = Path("/ceph/dwong/work/test/meta_energy_10.h5")  # adjust
attrs, meta_df = read_meta_h5(meta_path)
print(attrs)           # {'n_channels': ..., 'trace_samples': ..., ...}
meta_df.head()
```
