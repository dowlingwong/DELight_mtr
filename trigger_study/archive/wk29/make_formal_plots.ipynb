```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
import h5py

# --- Settings ---
base_root = Path("/ceph/dwong/work/threshold/pink_run1")
noise_powers = [15.4, 42.8, 126.7, 472.0]

A_SIGMA = 0.162217  # fallback: sigma(noise_power) ≈ A * sqrt(noise_power)

# We always want 3σ, 4σ, 5σ
sigma_multipliers = [5, 6, 7]
mult_labels = {5: "5σ", 6: "6σ", 7: "7σ"}
energy_re = re.compile(r"amplitudes_energy_(?P<E>[-+]?\d+(?:\.\d+)?)\.h5$")


def load_amplitudes(path: Path) -> np.ndarray:
    with h5py.File(path, "r") as f:
        if "amplitudes" not in f:
            raise KeyError(f"Missing 'amplitudes' dataset in {path}")
        data = np.asarray(f["amplitudes"][:])
    if data.ndim == 1:
        data = data[:, None]
    if data.ndim != 2:
        raise ValueError(
            f"Unexpected shape in {path}: {data.shape} (need (repeats, n_channels))"
        )
    return data


def sigma_from_positive_noise(noise_power: float) -> tuple[float, str]:
    """Match the later Gaussian-fit cell: std of positive amplitudes in energy=0 file."""
    path = base_root / f"noise_{noise_power}" / "fit_with_shift_amp" / "amplitudes_energy_0.h5"
    if path.exists():
        try:
            flat = load_amplitudes(path).ravel().astype(float)
            pos = flat[flat > 0]
            if pos.size > 0:
                return float(np.std(pos, ddof=0)), "measured (pos amps)"
            else:
                print(f"[warn] No positive amplitudes in {path}, using formula.")
        except Exception as exc:
            print(f"[warn] Failed reading {path}: {exc}; using formula.")
    else:
        print(f"[warn] Missing {path}, using formula.")
    return float(A_SIGMA * np.sqrt(noise_power)), "formula"


# Precompute sigmas the same way as the Gaussian-fit cell
noise_sigmas = {p: sigma_from_positive_noise(p) for p in noise_powers}

for noise_power in noise_powers:
    base_dir = base_root / f"noise_{noise_power}" / "fit_with_shift_amp"
    if not base_dir.exists():
        print(f"[warn] Missing noise directory: {base_dir}")
        continue

    files = []
    for f in base_dir.glob("amplitudes_energy_*.h5"):
        m = energy_re.match(f.name)
        if m:
            files.append((float(m.group("E")), f))
    files.sort(key=lambda t: t[0])

    if not files:
        print(f"[warn] No amplitude files found in {base_dir}")
        continue

    sigma, sigma_src = noise_sigmas[noise_power]
    thresholds = {m: m * sigma for m in sigma_multipliers}

    print(f"\n=== Noise power {noise_power} ===")
    print(f"  σ ≈ {sigma:.4f} ({sigma_src})")
    for m in sigma_multipliers:
        print(f"  {m}σ threshold ≈ {thresholds[m]:.4f}")

    pass_rates = {m: [] for m in sigma_multipliers}
    available_E = []
    n_channels = None

    for E, path in files:
        try:
            data = load_amplitudes(path)
        except Exception as exc:
            print(f"[warn] Failed to load {path}: {exc}")
            continue

        if n_channels is None:
            n_channels = data.shape[1]
        elif data.shape[1] != n_channels:
            print(f"[warn] Channel count changed {n_channels} -> {data.shape[1]} for {path}")
            n_channels = data.shape[1]

        available_E.append(E)

        for m in sigma_multipliers:
            thr = thresholds[m]
            passes = (data >= thr).any(axis=1)  # trace passes if any channel exceeds thr
            pass_rates[m].append(passes.mean())

    if not available_E:
        print(f"[warn] No energies found for noise={noise_power}, skipping plot.")
        continue

    plt.figure(figsize=(6.5, 4.5))
    for m in sigma_multipliers:
        plt.plot(available_E, pass_rates[m], label=mult_labels[m])

    plt.xlabel("Energy (eV)")
    plt.ylabel("Pass rate")
    plt.title(f"Pass rate vs Energy (noise = {noise_power})")
    plt.ylim(0.0, 1.05)
    plt.grid(True, which="both", linestyle="--", alpha=0.4)
    plt.legend(title="Thresholds")
    plt.tight_layout()
    plt.show()

    print("\nPass rates:")
    header = "Energy".ljust(8) + " " + "  ".join([f"{mult_labels[m]:>10}" for m in sigma_multipliers])
    print(header)
    for i, E in enumerate(available_E):
        row = [f"{E:<8}"]
        for m in sigma_multipliers:
            row.append(f"{pass_rates[m][i]:>10.4f}")
        print(" ".join(row))
    if n_channels is not None:
        print(f"Channels per trace: {n_channels}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import h5py

# Gaussian sigma from positive-only noise amplitudes (energy=0)
base_root = Path("/ceph/dwong/work/threshold/pink_run1")
noise_powers = [15.4, 42.8, 126.7, 472.0]

for noise_power in noise_powers:
    path = base_root / f"noise_{noise_power}" / "fit_with_shift_amp" / "amplitudes_energy_0.h5"
    if not path.exists():
        print(f"[warn] Missing noise= {noise_power} file: {path}")
        continue

    with h5py.File(path, "r") as f:
        if "amplitudes" not in f:
            print(f"[warn] No 'amplitudes' dataset in {path}")
            continue
        amps = np.asarray(f["amplitudes"][:])

    flat = amps.ravel().astype(float)
    pos = flat[flat > 0]
    if pos.size == 0:
        print(f"[warn] No positive amplitudes for noise={noise_power}")
        continue

    mu = float(np.mean(pos))
    sigma = float(np.std(pos, ddof=0))  # MLE sigma on positive side

    print(f"noise={noise_power}: mu≈{mu:.4f}, sigma≈{sigma:.4f}, samples={pos.size}")

    plt.figure(figsize=(6, 4))
    n, bins, _ = plt.hist(pos, bins=50, density=True, alpha=0.6, color="steelblue", label="data (+)")

    xs = np.linspace(bins[0], bins[-1], 200)
    pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((xs - mu) / sigma) ** 2)
    plt.plot(xs, pdf, "r-", label=f"Gaussian fit\nσ≈{sigma:.4f}")

    plt.title(f"Positive noise-only amplitudes, noise={noise_power}")
    plt.xlabel("Amplitude")
    plt.ylabel("Density")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
